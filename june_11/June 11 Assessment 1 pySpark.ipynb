```python
#61
from pyspark.sql.window import Window
from pyspark.sql.functions import rank, sum

patient_spending = appointments_df.groupBy(
    "patient_id"
).agg(
    sum("consultation_fee").alias("total_fee")
)

window_spec = Window.orderBy(
    patient_spending.total_fee.desc()
)

patient_spending.withColumn(
    "rank",
    rank().over(window_spec)
).show()
```

```python
#62
from pyspark.sql.functions import dense_rank

patient_spending.withColumn(
    "dense_rank",
    dense_rank().over(window_spec)
).show()
```

```python
#63
from pyspark.sql.functions import row_number

patient_spending.withColumn(
    "row_number",
    row_number().over(window_spec)
).show()
```

```python
#64
patient_spending.withColumn(
    "rank",
    rank().over(window_spec)
).filter(
    "rank = 1"
).show()
```

```python
#65
patient_spending.show(3)
```

```python
#66
from pyspark.sql.functions import rank

patient_city = patients_df.join(
    patient_spending,
    on="patient_id",
    how="inner"
)

city_window = Window.partitionBy(
    "city"
).orderBy(
    patient_city.total_fee.desc()
)

patient_city.withColumn(
    "rank",
    rank().over(city_window)
).filter(
    "rank = 1"
).show()
```

```python
#67
city_window_low = Window.partitionBy(
    "city"
).orderBy(
    patient_city.total_fee.asc()
)

patient_city.withColumn(
    "rank",
    rank().over(city_window_low)
).filter(
    "rank = 1"
).show()
```

```python
#68
running_window = Window.orderBy(
    "total_fee"
)

from pyspark.sql.functions import sum

patient_spending.withColumn(
    "running_total",
    sum("total_fee").over(running_window)
).show()
```

```python
#69
from pyspark.sql.functions import lead

patient_spending.withColumn(
    "next_fee",
    lead("total_fee").over(window_spec)
).show()
```

```python
#70
from pyspark.sql.functions import lag

patient_spending.withColumn(
    "previous_fee",
    lag("total_fee").over(window_spec)
).show()
```

```python
#93
patients_df = patients_df.na.fill({
    "city":"Unknown",
    "blood_group":"Not Available"
})

appointments_df = appointments_df.na.fill({
    "consultation_fee":0
})

flat_pref_df = flat_pref_df.na.fill({
    "phone":"Not Provided",
    "email":"Not Provided"
})
```

```python
#94
final_df = patients_df.join(
    appointments_df,
    on="patient_id",
    how="left"
).join(
    flat_pref_df,
    on="patient_id",
    how="left"
)

final_df.show()
```

```python
#95
from pyspark.sql.functions import when,col

final_df = final_df.withColumn(
    "age_group",
    when(col("age") < 30,"Young")
    .when(col("age") < 50,"Adult")
    .otherwise("Senior")
)

final_df.show()
```

```python
#96
from pyspark.sql.functions import sum

final_df.agg(
    sum("consultation_fee")
    .alias("total_revenue")
).show()
```
