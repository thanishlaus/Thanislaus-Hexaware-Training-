```python
import pandas as pd

products_df = pd.read_csv("inventory_products.csv")
warehouse_df = pd.read_csv("warehouse_status.csv")

print(products_df)
print()
print(warehouse_df)
```

```python
inventory_df = pd.DataFrame({
    "product_id":[101,102,103],
    "warehouse_id":[1,1,2]
})

inventory_df
```

```python
master_df = pd.merge(
    inventory_df,
    products_df,
    on="product_id"
)

master_df
```

```python
master_df = pd.merge(
    master_df,
    warehouse_df,
    on="warehouse_id"
)

master_df
```

```python
master_df["reorder_flag"] = master_df["stock_status"].apply(
    lambda x: "YES" if x=="Understocked" else "NO"
)

master_df
```

```python
master_df.to_csv(
    "master_inventory.csv",
    index=False
)

print("File generated successfully")
```

```python
import pandas as pd
import numpy as np

# Load data
df = pd.read_csv("stock_movements.csv")

# Calculate stock levels
stock_summary = (
    df.groupby(
        ["product_id",
         "product_name",
         "reorder_level"]
    )["quantity"]
    .sum()
    .reset_index()
)

stock_summary.rename(
    columns={"quantity":"current_stock"},
    inplace=True
)

# Create reorder flag
stock_summary["status"] = np.where(
    stock_summary["current_stock"] <
    stock_summary["reorder_level"],
    "REORDER",
    "OK"
)

# Filter low stock items
low_stock = stock_summary[
    stock_summary["status"]=="REORDER"
]

# Save report
low_stock.to_csv(
    "low_stock_report.csv",
    index=False
)

print("Low Stock Report Generated Successfully")
print()
print(low_stock)
```
