```python
from pyspark.sql.functions import col

preferences_df = spark.read \
.option("multiline","true") \
.json("/Volumes/hexa_databrick_7405612841012936/default/retail-volume/customer_preferences.json")

preferences_df.show()

preferences_df.printSchema()
```

```python
preferences_flat_df = preferences_df.select(
    "customer_id",
    "preferred_channel",
    col("loyalty.tier").alias("tier"),
    col("loyalty.points").alias("points")
)

preferences_flat_df.show()
```

```python
spark.read.format("delta") \
.load("/Volumes/hexa_databrick_7405612841012936/default/retail-volume/preferences") \
.show()
```

```python
preferences_flat_df.write \
.mode("overwrite") \
.format("delta") \
.save("/Volumes/hexa_databrick_7405612841012936/default/retail-volume/preferences")
```
