```python
#80
freq = rdd.countByValue()

print("===== LEARNING PLATFORM SUMMARY =====")

print("Total Records:",
      rdd.count())

print("Unique Activities:",
      rdd.distinct().count())

print("Most Common Activity:",
      max(freq.items(),
          key=lambda x: x[1]))

print("Least Common Activity:",
      min(freq.items(),
          key=lambda x: x[1]))

print("LOGIN Percentage:",
      (rdd.filter(
          lambda x: x == "LOGIN"
       ).count() / rdd.count()) * 100)

print("VIDEO_PLAY Percentage:",
      (rdd.filter(
          lambda x: x == "VIDEO_PLAY"
       ).count() / rdd.count()) * 100)
```
