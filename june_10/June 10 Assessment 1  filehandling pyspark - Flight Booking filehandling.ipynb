```python
#15
flights_df.orderBy(
    flights_df.ticket_price.desc()
).show()
```

```python
#16
from pyspark.sql.functions import col

flights_df = flights_df.withColumn(
    "tax",
    col("ticket_price") * 0.05
)

flights_df.show()
```

```python
#17
flights_df = flights_df.withColumn(
    "final_price",
    col("ticket_price") + col("tax")
)

flights_df.show()
```

```python
#18
from pyspark.sql.functions import when

flights_df = flights_df.withColumn(
    "price_category",
    when(col("ticket_price") >= 30000, "Premium")
    .when(col("ticket_price") >= 10000, "Standard")
    .otherwise("Budget")
)

flights_df.show()
```

```python
#19
flights_df.groupBy(
    "price_category"
).count().show()
```

```python
#20
flights_df.filter(
    flights_df.booking_status == "Confirmed"
).write.mode("overwrite").parquet(
    "confirmed_flights.parquet"
)
```
