```python
bookings_df = spark.read.csv(
    "bookings.csv",
    header=True,
    inferSchema=True
)
df.show()
```

```python
from pyspark.sql.functions import col
bookings_df.filter(
    col("city").isNull()
).show()

bookings_df.filter(
    col("provider").isNull()
).show()
```

```python
drop_any_null_df = booking_df.na.drop()
drop_any_null_df.show()
```

```python
clean_df = booking_df.na.drop(
    subset=[
        "customer_name",
        "service_type",
        "booking_amount"
    ]
)
clean_df.show()
```

```python
filled_df=booking_df.na.fill(
    {"city":"Unknown"}
)
filled_df.show()
```

```python
filled_df=booking_df.na.fill(
    {
        "provider":"Not Available",
        "booking_status":"Not Provided",
        "payment_mode":"Not provided",
        "booking_amount":0
    }
)
filled_df.show()
```
